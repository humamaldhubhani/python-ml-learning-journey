```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder

# Sample data
data = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'red', 'blue'],
    'size': ['S', 'M', 'L', 'M', 'S'],
    'price': [10, 20, 30, 25, 15]
})

print("Original data:")
print(data)

# Initialize the encoder
encoder = OneHotEncoder(sparse_output=False)  # sparse_output=False returns dense array

# Select categorical columns to encode
categorical_columns = ['color', 'size']

# Fit and transform the data
encoded_array = encoder.fit_transform(data[categorical_columns])

# Get feature names
feature_names = encoder.get_feature_names_out(categorical_columns)

# Create DataFrame with encoded features
encoded_df = pd.DataFrame(encoded_array, columns=feature_names)

print("\nEncoded features:")
print(encoded_df)
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

# Sample data
data = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'red', 'blue'],
    'size': ['S', 'M', 'L', 'M', 'S'],
    'price': [10, 20, 30, 25, 15]
})

print("Original data:")
print(data)

# Create a copy of the data
data_encoded = data.copy()

# Encode multiple columns
encoders = {}
categorical_columns = ['color', 'size']

for col in categorical_columns:
    encoders[col] = LabelEncoder()
    data_encoded[f'{col}_encoded'] = encoders[col].fit_transform(data[col])

print("\nData with all categorical columns label encoded:")
print(data_encoded)

# Show mappings
for col in categorical_columns:
    print(f"\n{col} mapping: {dict(zip(encoders[col].classes_, encoders[col].transform(encoders[col].classes_)))}")
```
