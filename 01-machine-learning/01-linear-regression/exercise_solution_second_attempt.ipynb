```python
import pandas as pd
import numpy as nm
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv("canada_per_capita_income.csv")
df.head(5)
```

```python
df.rename(columns={"per capita income (US$)":"income"},inplace=True)
df.head(5)
```

```python
plt.xlabel("Year")
plt.ylabel("per capita income (US$)")
plt.scatter(df[["year"]],df.income,marker="+")
```

```python
reg=linear_model.LinearRegression()
reg.fit(df[["year"]],df.income)
```

```python
print("Intercept : ", reg.intercept_)
print("Coefficient : ", reg.coef_)
```

```python
reg.predict([[2016]])
```

```python
(828.46507522*2016)+(-1632210.7578554575)
```

```python
df2=pd.read_csv("years.csv")
df2.info()
```

```python
prediction = reg.predict(nm.array(df2[["years"]]))
prediction
```

```python
df2.head(5)
```

```python
df2["predicted income"]=prediction
df2.head(5)
```

```python
df2.to_csv("predictions.csv",index=False)
```
