```python
%%timeit -n1 -r1
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
(X_train , y_train) , (X_test , y_test) = datasets.mnist.load_data()
X_train.shape , X_test.shape , y_train.shape , y_test.shape
```

```python
# scale Xs for better accuracy by converting the pexel range from (0 , 255) to (0,1")
X_train = X_train / 255
X_test = X_test / 255
```

```python
X_train = X_train.reshape(len(X_train) , 28* 28)
X_test  = X_test.reshape(len(X_test) , 28* 28)
```

```python
X_train.shape , X_test.shape , y_train.shape , y_test.shape
```

# Simple Nueral Network with no convloutional layers

```python
model = models.Sequential([
    layers.Dense( 1000 , input_shape = (28*28 , ) , activation='relu'),
    layers.Dense( 1000 , activation='relu'),
    layers.Dense( 1000 , activation='sigmoid'),
    layers.Dense( 10 , activation='sigmoid')
])

model.compile(
    optimizer="adam",
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
%%timeit -n1 -r1
model.fit(X_train , y_train , epochs=10)
```
